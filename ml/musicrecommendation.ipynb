```python


songs = df.copy()

songs = songs.drop(columns=["Unnamed: 0"], errors="ignore")

songs = songs.dropna(
    subset=["track_name", "artists", "track_genre"]
)
songs["album_name"] = songs["album_name"].fillna("Unknown Album")


for col in ["track_name", "artists", "album_name", "track_genre"]:
    songs[col] = songs[col].astype(str).str.strip()


songs = songs[
    (songs["track_name"] != "") &
    (songs["artists"] != "") &
    (songs["track_genre"] != "")
]


songs = songs.drop_duplicates()


print("Cleaning complete!")
print("Original dataset:", df.shape)
print("Cleaned dataset:", songs.shape)
print("Missing values remaining:")
print(songs.isnull().sum())
```

```python
songs = pd.read_csv("vayam_cleaned_songs.csv")

print("Dataset loaded successfully!")
print("Shape:", songs.shape)
display(songs.head())
```

```python

audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo"
]

audio_features = [
    col for col in audio_features if col in songs.columns
]

print("Audio features selected:")
print(audio_features)

print("\nGenre column available:", "track_genre" in songs.columns)
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import csr_matrix, hstack

# Missing numeric values ko median se fill karo
X_audio = songs[audio_features].copy()
X_audio = X_audio.fillna(X_audio.median())

scaler = StandardScaler()
audio_scaled = scaler.fit_transform(X_audio)

genre_vectorizer = TfidfVectorizer()
genre_matrix = genre_vectorizer.fit_transform(
    songs["track_genre"].fillna("").astype(str)
)

X = hstack([
    csr_matrix(audio_scaled),
    genre_matrix
], format="csr")

print("Feature preparation complete!")
print("Number of songs:", X.shape[0])
print("Number of features:", X.shape[1])
```

```python

from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(
    n_neighbors=6,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

model.fit(X)

print("Vayam Music recommendation model is ready!")
```

```python

def recommend_songs(song_name, number_of_recommendations=5):

    matches = songs[
        songs["track_name"].str.contains(
            song_name, case=False, na=False, regex=False
        )
    ]

    if matches.empty:
        print("Song not found. Try another song name.")
        return

    song_index = matches.index[0]

    position = songs.index.get_loc(song_index)

    distances, indices = model.kneighbors(
        X[position],
        n_neighbors=number_of_recommendations + 1
    )

    
    recommended_positions = [
        i for i in indices[0] if i != position
    ][:number_of_recommendations]

    result = songs.iloc[recommended_positions][
        ["track_name", "artists", "track_genre"]
    ]

    print("Recommendations for:", songs.iloc[position]["track_name"])
    display(result.reset_index(drop=True))

```

** MOOD BASED MUSIC RECOMMENDATION SYSTEM**

```python

def recommend_by_mood(mood, number=10):

    mood_targets = {
        "happy": {
            "valence": 0.80,
            "energy": 0.75,
            "danceability": 0.75
        },
        "sad": {
            "valence": 0.20,
            "energy": 0.35,
            "danceability": 0.40
        },
        "relaxed": {
            "valence": 0.55,
            "energy": 0.25,
            "danceability": 0.45
        },
        "energetic": {
            "valence": 0.65,
            "energy": 0.90,
            "danceability": 0.70
        },
        "focus": {
            "valence": 0.50,
            "energy": 0.35,
            "danceability": 0.40
        }
    }

    mood = mood.lower().strip()

    if mood not in mood_targets:
        print("Choose: happy, sad, relaxed, energetic, focus")
        return

    targets = mood_targets[mood]

    # Dataset ke har song ka mood se difference calculate karo
    distances = sum(
        (songs[col].fillna(songs[col].median()) - target) ** 2
        for col, target in targets.items()
    )

    result = songs.loc[distances.nsmallest(number).index, [
        "track_name", "artists", "track_genre",
        "energy", "valence", "danceability"
    ]]

    display(result.reset_index(drop=True))


# Test mood recommendations
recommend_by_mood("SAD")
```

```python
recommend_by_mood("happy")
recommend_by_mood("sad")
recommend_by_mood("relaxed")
recommend_by_mood("energetic")
recommend_by_mood("focus")
```

```python
MODEL_DIR = Path("models")

artifacts = joblib.load(
    MODEL_DIR / "vayam_recommender.pkl"
)

loaded_model = artifacts["similar_song_model"]
loaded_scaler = artifacts["scaler"]
loaded_vectorizer = artifacts["genre_vectorizer"]
loaded_audio_features = artifacts["audio_features"]

loaded_songs = pd.read_csv(
    MODEL_DIR / "vayam_cleaned_songs.csv"
)

print("Model loaded:", loaded_model is not None)
print("Songs loaded:", len(loaded_songs))
print("Audio features:", loaded_audio_features)
print("Model verification successful!")
```

```python
from scipy.sparse import csr_matrix, hstack

X_audio_loaded = loaded_songs[loaded_audio_features].copy()
X_audio_loaded = X_audio_loaded.fillna(
    X_audio_loaded.median()
)

audio_scaled_loaded = loaded_scaler.transform(
    X_audio_loaded
)

genre_matrix_loaded = loaded_vectorizer.transform(
    loaded_songs["track_genre"].fillna("").astype(str)
)

X_loaded = hstack(
    [csr_matrix(audio_scaled_loaded), genre_matrix_loaded],
    format="csr"
)

def recommend_saved_model(song_name, number=5):
    matches = loaded_songs[
        loaded_songs["track_name"].str.contains(
            song_name, case=False, na=False, regex=False
        )
    ]

    if matches.empty:
        print("Song not found. Try another song name.")
        return

    position = loaded_songs.index.get_loc(matches.index[0])

    distances, indices = loaded_model.kneighbors(
        X_loaded[position],
        n_neighbors=number + 1
    )

    recommendations = [
        i for i in indices[0] if i != position
    ][:number]

    display(
        loaded_songs.iloc[recommendations][
            ["track_name", "artists", "track_genre"]
        ].reset_index(drop=True)
    )

# Test
recommend_saved_model("Gerua", 5)
```
